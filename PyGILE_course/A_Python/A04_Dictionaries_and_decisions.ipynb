## 1. Dictionaries: looking things up by name

A **dictionary** (`dict` in Python) stores pairs of values. Each pair has a **key**, the name you look up, and a **value**, what you find under it. It works like a real dictionary: you look up a word and find its meaning.

You write a dictionary with curly brackets `{ }`. Each pair is written `key: value`, and commas separate the pairs.

To look up a value, put its key in square brackets: `peak_heights_m["Makalu"]`. Each key can appear only once. Keys are usually strings; values can be of any type.

````python
peak_heights_m = {
    "Everest": 8849,
    "Kanchenjunga": 8586,
    "Lhotse": 8516,
    "Makalu": 8485,
    "Cho Oyu": 8188,
    "Dhaulagiri I": 8167,
    "Manaslu": 8163,
    "Annapurna I": 8091,
}
print(peak_heights_m["Makalu"])
print(len(peak_heights_m))
````

````python
everest = {"name": "Everest", "height_m": 8849, "first_ascent": 1953, "province": "Koshi"}
print(everest["province"])
print(everest["height_m"] - 8000)
````

The first dictionary has 8 pairs, one per peak, so `len()` gives 8. You no longer need to know that Makalu is at index 3: the name is enough.

The second dictionary describes **one** peak. Its keys are the kinds of information, like `"province"`. A dictionary like this is called a **record**; step 7 puts many records together.

## 2. Adding, updating and checking keys

You can change a dictionary after you make it, with the same square brackets:

- `d[key] = value` with a **new** key adds a new pair.
- `d[key] = value` with a key that **already exists** replaces its value.

There is no warning when a value is replaced, so check your spelling.

`key in d` asks whether a key exists, and gives `True` or `False`. It looks at the keys only, not at the values.

**Predict:** look at the last line of the cell below. Will it print `True` or `False`?

````python
ascent_years = {"Everest": 1953, "Makalu": 1959}    # Makalu typed wrongly
ascent_years["Makalu"] = 1955     # existing key: the value is replaced
ascent_years["Lhotse"] = 1956     # new key: a new pair is added
print(ascent_years)
print("Lhotse" in ascent_years)
print(1956 in ascent_years)
````

The last line prints `False`. 1956 is a *value* in the dictionary, but `in` checks only the keys, which here are peak names.

## 3. Looping over a dictionary

A dictionary has three methods that give you its contents:

| Method | Gives |
| --- | --- |
| `.keys()` | all the keys |
| `.values()` | all the values |
| `.items()` | all the pairs, as (key, value) |

Loop over `.items()` with two loop variables, one for the key and one for the value. This works like `enumerate()` in Lesson A3.

A `for` loop over the dictionary itself gives the keys. `sum()`, `min()` and `max()` work on `.values()`.

````python
print(peak_heights_m.keys())
print(sum(peak_heights_m.values()))
````

````python
for name, height_m in peak_heights_m.items():
    print(f"{name}: {height_m:,} m")
````

`.keys()` prints as `dict_keys([...])`. That is Python's way of saying "these are the keys of a dictionary"; you can loop over it like a list. The heights add up to 67,045 m, as in Lesson A3.

## 4. Decisions: `if` and `else`

Until now, every line of your code ran. With `if`, a block runs **only when** a condition is `True`:

```python
if condition:
    lines that run when the condition is True
else:
    lines that run when it is False
```

The **condition** is any expression that gives `True` or `False`, such as the comparisons from Lesson A2. Like a `for` loop, `if` and `else` end with a colon, and their blocks are indented. The `else` part is optional.

**Predict:** which message will the first cell print?

````python
altitude_m = 3440     # Namche Bazaar
if altitude_m > 3000:
    print("Above 3,000 m: watch for signs of altitude sickness.")
else:
    print("Below 3,000 m.")
````

````python
for name, height_m in peak_heights_m.items():
    if height_m > 8500:
        print(name, "is above 8,500 m")
````

3,440 is greater than 3,000, so the first message prints and the `else` block is skipped.

In the second cell, the `if` sits inside a loop. Its block is indented twice: once for the loop, once for the `if`. Only Everest, Kanchenjunga and Lhotse pass the test.

Decisions take a little more room than earlier code. From here on, **Q1** still needs one line, but **Q2** and **Q3** answers may be a few lines longer than their labels.

## 5. More than two choices: `elif`

When there are more than two possibilities, add one or more `elif` blocks between `if` and `else`. `elif` is short for "else if".

Python checks the conditions **from top to bottom**. It runs the block of the **first** condition that is `True`, then skips all the rest. If none is `True`, the `else` block runs.

Because only the first match counts, the order matters. Put the strictest test first.

The cell below sorts a place into an elevation zone. The zone names and limits are made up for practice. **Predict:** which zone will Jomsom (2,720 m) get?

````python
elevation_m = 2720     # Jomsom
````

````python
if elevation_m >= 5000:
    zone = "high mountain"
elif elevation_m >= 2500:
    zone = "mountain"
elif elevation_m >= 1000:
    zone = "hill"
else:
    zone = "lowland"

print(zone)
````

The first test, `>= 5000`, is `False`. The second, `>= 2500`, is `True`, so `zone` becomes `"mountain"` and the last two tests are never checked.

## 6. Combining conditions: `and`, `or`, `not`

Three words combine or flip conditions:

| Word | Gives `True` when | Example | Result |
| --- | --- | --- | --- |
| `and` | **both** sides are `True` | `8163 > 8100 and 8163 < 8200` | `True` |
| `or` | **at least one** side is `True` | `8163 > 8500 or 8163 < 8200` | `True` |
| `not` | the condition is `False` | `not 8163 > 8500` | `True` |

Each side must be a full comparison. Write `height_m > 8100 and height_m < 8200`, not `height_m > 8100 and < 8200`.

`not in` is a handy short form: `"Bagmati" not in peaks_per_province` means `not ("Bagmati" in peaks_per_province)`.

````python
height_m = 8163       # Manaslu
province = "Gandaki"

print(height_m > 8100 and province == "Gandaki")
print(height_m > 8500 and province == "Gandaki")
print(height_m > 8500 or province == "Gandaki")
print(not province == "Koshi")
````

````python
for name, height_m in peak_heights_m.items():
    if height_m > 8400 and height_m < 8600:
        print(name)
````

The four lines print `True`, `False`, `True`, `True`. The second is `False` because Manaslu is not above 8,500 m, and `and` needs both sides.

The loop prints the peaks between 8,400 m and 8,600 m: Kanchenjunga, Lhotse and Makalu.

## 7. A table of records: a list of dictionaries

Real data usually looks like a table: one **row** per thing, one **column** per kind of information. In plain Python, you can build a table as a list of dictionaries:

- each dictionary is one **record**, one row of the table;
- each key is a column name, like `"height_m"`;
- the list holds all the rows in order.

Get a row with a list index, then a value with a key: `peak_records[0]["name"]`. In a loop, each record arrives in turn, and you read its values by key.

This is also how the course files will look when you read them in Lesson A7. In Lesson A9, pandas tables work on the same idea.

````python
peak_records = [
    {"name": "Everest",      "height_m": 8849, "first_ascent": 1953, "province": "Koshi"},
    {"name": "Kanchenjunga", "height_m": 8586, "first_ascent": 1955, "province": "Koshi"},
    {"name": "Lhotse",       "height_m": 8516, "first_ascent": 1956, "province": "Koshi"},
    {"name": "Makalu",       "height_m": 8485, "first_ascent": 1955, "province": "Koshi"},
    {"name": "Cho Oyu",      "height_m": 8188, "first_ascent": 1954, "province": "Koshi"},
    {"name": "Dhaulagiri I", "height_m": 8167, "first_ascent": 1960, "province": "Gandaki"},
    {"name": "Manaslu",      "height_m": 8163, "first_ascent": 1956, "province": "Gandaki"},
    {"name": "Annapurna I",  "height_m": 8091, "first_ascent": 1950, "province": "Gandaki"},
]
print(len(peak_records))
print(peak_records[0])
print(peak_records[0]["name"], peak_records[-1]["name"])
````

````python
for peak in peak_records:
    print(f"{peak['name']} ({peak['province']}): {peak['height_m']:,} m, first climbed {peak['first_ascent']}")
````

Inside an f-string that uses double quotation marks, write the keys with single quotation marks: `peak['name']`. Otherwise Python thinks the string has ended.

The spaces inside the records only line up the columns for your eyes; Python ignores them.

## 8. Counting and filtering records

Two loop patterns answer most questions about a table.

**Counting:** how many records pass a test? Start a counter at `0`. In the loop, add 1 with `+=` when a record passes.

**Filtering:** which records pass a test? Start an empty list. In the loop, `.append()` each record, or one of its values, when it passes.

Both patterns combine a `for` loop with an `if` inside it. You can count and filter in the same loop.

````python
count = 0
for peak in peak_records:
    if peak["province"] == "Gandaki":
        count += 1
print("Peaks in Gandaki:", count)
````

````python
early_names = []
for peak in peak_records:
    if peak["first_ascent"] < 1955:
        early_names.append(peak["name"])
print(early_names)
````

Three peaks are in Gandaki. Three peaks were first climbed before 1955: Everest, Cho Oyu and Annapurna I.

```python
    else:
        zone = "lowland"
    print(f"{town} ({elevation_m:,} m): {zone}")
```

Kathmandu is `hill`, Pokhara `lowland`, Lukla `mountain` and Everest Base Camp `high mountain`. The `print` line is indented once, so it runs for every town, after the decision.

```python
stations = [
    {"name": "Station A", "district": "Kaski", "monthly_mm": [35, 59, 78, 151, 206, 536, 1045, 952, 592, 198, 42, 42]},
    {"name": "Station C", "district": "Mustang", "monthly_mm": [3, 5, 6, 11, 17, 37, 69, 68, 37, 14, 3, 3]},
```

## Recap

- A **dictionary** stores `key: value` pairs in `{ }`. Look up with `d[key]`; add or replace with `d[key] = value`.
- `key in d` checks the keys. `.keys()`, `.values()` and `.items()` let you loop over a dictionary.
- `if`, `elif` and `else` choose which block runs. Only the first `True` condition counts, so order matters.
- `and`, `or` and `not` combine conditions; each side must be a full comparison.
- A list of dictionaries is a small table of records. Count with `+= 1` and filter with `.append()` inside an `if`.

### Words you learned

| Word | Meaning |
| --- | --- |
| dictionary (`dict`) | A collection of `key: value` pairs in curly brackets |
| key | The name you look up in a dictionary, like `"Makalu"` |
| value | What is stored under a key, like `8485` |
| pair | One key together with its value |
| `.keys()`, `.values()`, `.items()` | Methods that give the keys, the values, or both |
| condition | An expression that gives `True` or `False`, used by `if` |
| `if` | Runs its block only when the condition is `True` |
| `elif` | "Else if": another condition, checked only if the earlier ones were `False` |
| `else` | Runs its block when no condition before it was `True` |
| `and`, `or`, `not` | Words that combine or flip conditions |
| record | A dictionary describing one thing; one row of a table |
| counter | A variable that starts at 0 and grows by 1 with `+= 1` |
| filter | To keep only the records that pass a test |
| `KeyError` | An error raised when a key is not in the dictionary |
