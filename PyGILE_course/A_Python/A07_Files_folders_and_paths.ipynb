## 1. Files, folders and the working directory

A **file** is a named piece of data stored on a disk: a notebook, a table, a photo. A **folder** (also called a **directory**) holds files and other folders. The course is organised like this:

```
PyGILE_course/
├── A_Python/      <- this notebook is here
├── data/          <- nepal_peaks.csv, practice_landslides.csv, ...
└── outputs/       <- files the lessons write (you create it in step 8)
```

A **path** is the address of a file: the folders to pass through, then the file name. An **absolute path** starts at the top of the disk, such as `/home/sita/PyGILE_course/data/nepal_peaks.csv` or `C:\Users\Sita\PyGILE_course\data\nepal_peaks.csv`. It works only on that one computer.

A **relative path** starts from the **working directory**: the folder Python is "standing in". In JupyterLab, that is the notebook's own folder. Two short names help: `.` means "this folder" and `..` means "one folder up". So `../data` means: go up from `A_Python` to `PyGILE_course`, then into `data`.

Python's `pathlib` module gives you `Path` objects for working with paths. The `/` operator joins parts of a path, on every operating system.

````python
from pathlib import Path

print(Path.cwd())   # cwd = current working directory
````

````python
DATA = Path("..") / "data"
peaks_file = DATA / "nepal_peaks.csv"

print(peaks_file)
print(peaks_file.exists())
````

`Path.cwd()` shows the absolute path of the working directory. It is different on every computer, but it should end in `A_Python`.

`peaks_file` is the relative path `../data/nepal_peaks.csv` (Windows shows it with `\` instead of `/`). `.exists()` checks whether something is really there: `True`. Because the course uses relative paths, the same code works on your computer and on mine.

## 2. The parts of a path

A `Path` can tell you its parts:

| Attribute | Meaning | For `../data/practice_landslides.csv` |
| --- | --- | --- |
| `.name` | the file name | `practice_landslides.csv` |
| `.suffix` | the **extension**, from the last dot | `.csv` |
| `.stem` | the name without the extension | `practice_landslides` |
| `.parent` | the folder that contains it | `../data` |

The **extension** tells people and programs what kind of file it is: `.csv` is a table stored as text, `.ipynb` a notebook, `.png` an image.

These attributes only read the text of the path. The file does not need to exist.

## 3. Looking inside a folder

`.iterdir()` gives every item inside a folder, one at a time, so you can loop over it. The order is not guaranteed, so wrap it in `sorted()` to get a fixed order.

`.glob("*.csv")` gives only the items whose names match a **pattern**. In a pattern, `*` means "any text". So `"*.csv"` matches every name ending in `.csv`, and `"practice_*"` every name starting with `practice_`.

````python
for path in sorted(DATA.iterdir()):
    print(path.name)
````

````python
csv_files = sorted(DATA.glob("*.csv"))
print(len(csv_files), "CSV files")
for path in csv_files:
    print(path.name)
````

The `data` folder holds four files. `README.md` comes first, because `sorted()` puts capital letters before small letters. The pattern `*.csv` found three of them.

Read `README.md` in JupyterLab (double-click it in the file browser): it says where each data file comes from. Checking that is a good habit with any dataset.

## 4. Reading a text file

A CSV file is plain text. You could open it in any text editor and read it yourself.

`path.read_text(encoding="utf-8")` reads the whole file into one string. The **encoding** tells Python how the letters are stored as bytes; `utf-8` is the modern standard, and it handles Nepali script too.

`.splitlines()` cuts that string into a list of lines. In a **CSV** (comma-separated values) file, the first line is the **header**, with the column names. Every other line is one **record**, with its values separated by commas.

````python
peaks_text = peaks_file.read_text(encoding="utf-8")
print(peaks_text)
````

````python
lines = peaks_text.splitlines()
print(len(lines))
print(lines[0])
print(lines[1])
````

The file has 9 lines: one header and 8 records, one per peak. Each line is still a single string, such as `'Everest,8849,1953,Koshi,27.9881,86.9250'`.

## 5. Reading a CSV table with the `csv` module

Splitting at commas works for simple files. It breaks when a value contains a comma itself, such as `"Kathmandu, Bagmati"` in quotation marks. The **`csv` module** handles these cases correctly.

`csv.DictReader` reads each record as a **dictionary**, using the header for the keys. The result is the same kind of table of records you built by hand in A4.

To give it the file, you open the file first:

- `open(path, encoding="utf-8", newline="")` opens the file for reading. The `newline=""` setting is what the `csv` module asks for.
- `with ... as f:` keeps the file open while the indented block runs, then closes it for you.
- `list(...)` collects all the records into one list.

````python
import csv

with open(landslides_file, encoding="utf-8", newline="") as f:
    landslides = list(csv.DictReader(f))

print(len(landslides))
print(landslides[0])
````

40 records, each a dictionary. The first is landslide `LS001` in Dolakha. Look closely: `'area_m2': '9800'` has quotation marks. Every value read from a text file is a **string**, until you convert it.

## 6. Turning text into numbers

Values read from a file are text. Adding two of them joins the text instead of adding the numbers.

**Predict:** both areas below are `'9800'`. What will `+` give?

````python
landslides[0]["area_m2"] + landslides[1]["area_m2"]
````

`'98009800'`: two strings joined end to end. To calculate, convert each value first, as in A2:

- `int()` for whole numbers, such as areas in whole square metres or years,
- `float()` for numbers with decimals, such as latitude and longitude.

A date like `'2023-08-18'` is text too. A slice `[:4]` takes its first four characters, the year, and `int()` makes it a number.

````python
total_area_m2 = 0
for record in landslides:
    total_area_m2 += int(record["area_m2"])

print(f"Total area: {total_area_m2:,} m²")
print(f"First record year: {int(landslides[0]['date'][:4])}")
````

The 40 landslides cover 187,250 m² in total. The first record is from 2023.

## 7. Counting and filtering records

The patterns from A4 now work on a real file:

- **Filtering:** loop over the records, test each one with `if`, and `append` the ones you want to a new list.
- **Counting by category:** use a dictionary. If the category is not a key yet, start it at 0; then add 1.

Here you filter the landslides in Dolakha district, then count the landslides for each trigger (the cause recorded for each landslide).

````python
dolakha = []
for record in landslides:
    if record["district"] == "Dolakha":
        dolakha.append(record)

print(len(dolakha), "landslides in Dolakha")
````

````python
count_by_trigger = {}
for record in landslides:
    trigger = record["trigger"]
    if trigger not in count_by_trigger:
        count_by_trigger[trigger] = 0
    count_by_trigger[trigger] += 1

print(count_by_trigger)
````

9 of the 40 records are in Dolakha. Rainfall is the most common trigger (29), then road cuts (8) and earthquakes (3). Remember these are practice values, not real statistics.

## 8. Writing files to the outputs folder

Results are worth saving, so you can share them or use them in the next lesson. Two rules keep your work safe:

- **Never overwrite the original data.** Write new files to a separate folder, `outputs`.
- **Create the folder from code**, so the notebook works on a fresh computer. `OUT.mkdir(exist_ok=True)` creates the folder; `exist_ok=True` means "no error if it is already there".

`path.write_text(text, encoding="utf-8")` writes a string to a file. It replaces anything that was in that file before. Inside a string, `\n` means "start a new line".

To write a CSV table, open a file with `"w"` (for *write*), make a `csv.writer`, and give it one list per row with `.writerow()`.

````python
OUT = Path("..") / "outputs"
OUT.mkdir(exist_ok=True)

summary_file = OUT / "landslide_summary.txt"
summary_text = f"Landslide records: {len(landslides)}\nIn Dolakha: {len(dolakha)}\n"
summary_file.write_text(summary_text, encoding="utf-8")

print(summary_file.read_text(encoding="utf-8"))
````

````python
counts_file = OUT / "landslides_by_trigger.csv"

with open(counts_file, "w", encoding="utf-8", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["trigger", "count"])
    for trigger, count in count_by_trigger.items():
        writer.writerow([trigger, count])

print(counts_file.read_text(encoding="utf-8"))
````

Reading each file back is a quick check that it was written as you meant. You will also find both files in the `outputs` folder in JupyterLab's file browser. The CSV file can be opened in a spreadsheet program too.

def read_records(path):
    """Return the rows of a CSV file as a list of dictionaries, or [] if it is missing."""
    try:
        with open(path, encoding="utf-8", newline="") as f:
            return list(csv.DictReader(f))
    except FileNotFoundError:
        print(f"Missing file: {path}")

## Recap

- A **path** is a file's address. Relative paths like `Path("..") / "data"` start from the **working directory**, `Path.cwd()`.
- `pathlib` gives `.exists()`, `.name`, `.suffix`, `.stem`, `.parent`, `.iterdir()` and `.glob("*.csv")`.
- `read_text()` and `.splitlines()` read a text file; `csv.DictReader` reads a CSV into a list of dictionaries.
- Every value from a file is text: convert with `int()` or `float()` before calculating or comparing numbers.
- Write results to `../outputs` with `write_text()` or `csv.writer`, never over the original data.
