## 1. Your first function

You have already used functions that other people wrote: `print()`, `round()`, `len()` and `sum()`. A **function** is a named block of code that you can run whenever you need it.

You create your own function with the word `def` (short for *define*):

```
def function_name(parameter):
    body: the indented lines that do the work
```

The `def` line ends with a colon, like `for` and `if`. The indented lines below it are the **body**. Defining a function does not run it. It only stores the recipe under a name.

To run the recipe, you **call** the function: write its name followed by brackets, with a value inside.

````python
def describe_height(height_m):
    print(f"Height: {height_m:,} m, or {height_m / 1000} km")

describe_height(8849)   # Everest
describe_height(8586)   # Kanchenjunga
````

The `def` part stored the recipe. Each call then ran the body once. In the first call, the value `8849` went into the name `height_m`; in the second call, `8586` did.

The name in the `def` line, `height_m`, is a **parameter**: a placeholder. The value you put in the brackets when you call the function is an **argument**.

## 2. Several parameters

A function can take several parameters, separated by commas. When you call it, the arguments are matched to the parameters **by position**: the first argument goes into the first parameter, the second into the second, and so on.

This means the order matters. If you swap two arguments, Python does not warn you. It calculates with the wrong values.

**Predict:** the function below prints the gradient of a road. Before you run it, write down what you expect from each of the two calls.

````python
def print_gradient(rise_m, run_m):
    print(f"Gradient: {rise_m / run_m * 100:.1f} %")

print_gradient(150, 2000)   # climbs 150 m over 2 km
print_gradient(2000, 150)   # the same numbers, swapped
````

````python
def describe_peak(name, height_m):
    print(f"{name}: {height_m:,} m")

describe_peak("Makalu", 8485)
````

The first call gives the correct 7.5 %. The second call put 2,000 into `rise_m` and 150 into `run_m`, and gave a meaningless 1,333.3 %. Good parameter names, with units, make this kind of mistake easier to spot.

## 3. `return`: handing a value back

The functions so far only print. Printing shows a value on the screen, and then the value is gone: you cannot store it or calculate with it.

`return` hands a value back to the code that called the function. You can store the returned value in a variable, print it, or use it in a bigger calculation. `return` also ends the function: any lines after it in the body do not run.

A good habit: let functions **return** results, and do the printing outside them.

````python
def m_to_ft(length_m):
    return length_m * 3.28084

everest_ft = m_to_ft(8849)
print(round(everest_ft))
print(round(m_to_ft(8849) - m_to_ft(8586)))
````

`m_to_ft(8849)` is replaced by the value it returns, so `everest_ft` holds about 29,032 ft. In the last line, the two calls are worked out first, then subtracted: Everest is about 863 ft higher than Kanchenjunga.

**Predict:** the next function prints instead of returning. What do you think `result` will hold?

````python
def m_to_ft_printing(length_m):
    print(length_m * 3.28084)

result = m_to_ft_printing(8849)
print(result)
````

The function printed the feet value, but it handed nothing back. A function without `return` gives back `None`, Python's word for "no value". So `result` holds `None`, and you cannot do arithmetic with it.

One more thing: a variable created inside a function's body exists only while the function runs. Outside the function, Python does not know that name.

## 4. Default values and keyword arguments

A parameter can have a **default value**, written with `=` in the `def` line: `digits=1`. If the caller leaves that argument out, Python uses the default.

Parameters with defaults must come after the parameters without them.

When you call a function, you can also name an argument: `digits=2`. A named argument is a **keyword argument**. Keyword arguments can be given in any order, and they make a call easier to read.

````python
def gradient_percent(rise_m, run_m, digits=1):
    return round(rise_m / run_m * 100, digits)

print(gradient_percent(100, 3000))              # digits uses its default, 1
print(gradient_percent(100, 3000, digits=3))    # keyword argument
print(gradient_percent(run_m=1500, rise_m=240)) # any order when named
````

The first call used the default and gave 3.3 %. The second asked for three decimal places: 3.333 %. The third named both arguments, so their order did not matter: 16.0 %.

## 5. Docstrings: a label on your tool

A function is a tool that other people, and you in six months, will pick up and use. A **docstring** is a short description written as the first line of the body. It says what the function does, what the parameters mean, and what it returns.

A docstring sits inside triple quotes, `"""like this"""`. Triple quotes let a string run over several lines.

`help(function_name)` shows a function's docstring. This works for Python's own functions too.

````python
def area_ha(length_m, width_m):
    """Return the area of a rectangular plot in hectares.

    length_m and width_m are the two sides in metres.
    """
    return length_m * width_m / 10000

help(area_ha)
print(area_ha(120, 80))
````

`help` printed the function's name, its parameters and the docstring. The plot of 120 m by 80 m is 0.96 hectares.

## 6. Functions that call functions

A function's body can call other functions, including ones you wrote yourself. This is how bigger programs are built: many small tools, each doing one job well, combined into larger ones.

Small functions are easier to test. When `m_to_ft` is right, every function that uses it gets the right conversion for free.

Here `describe_in_feet` uses `m_to_ft` from step 3.

````python
def describe_in_feet(name, height_m):
    """Return a sentence giving a height in metres and in feet."""
    height_ft = round(m_to_ft(height_m))
    return f"{name}: {height_m:,} m ({height_ft:,} ft)"

print(describe_in_feet("Everest", 8849))
````

## 7. The `math` module: slopes in degrees and percent

Python comes with extra toolboxes called **modules**. `import math` loads the maths module. You then use its tools with a dot: `math.tan(...)`, `math.pi`.

Slopes can be measured in two ways. A **slope angle** in degrees is the angle between the ground and the horizontal. A **gradient** in percent is rise ÷ run × 100. They are linked by the tangent function:

- percent = tan(angle) × 100
- angle = atan(percent ÷ 100), where atan is the *inverse tangent*

One catch: `math` measures angles in **radians**, not degrees. A full circle is 360° or 2π radians. `math.radians()` converts degrees to radians, and `math.degrees()` converts back.

**Predict:** a 45° slope rises 1 m for every 1 m across. What gradient in percent is that?

````python
import math

print(math.pi)
print(math.tan(math.radians(45)))
````

````python
def degrees_to_percent(slope_deg):
    """Return a slope angle in degrees as a gradient in percent."""
    return math.tan(math.radians(slope_deg)) * 100

def percent_to_degrees(slope_percent):
    """Return a gradient in percent as a slope angle in degrees."""
    return math.degrees(math.atan(slope_percent / 100))

print(round(degrees_to_percent(45), 1))
print(round(degrees_to_percent(30), 1))
print(round(percent_to_degrees(7.5), 1))
````

`math.tan(math.radians(45))` gives 0.9999999999999999 instead of 1: the same tiny binary rounding you met in A1. So a 45° slope is a 100 % gradient. A 30° slope is already 57.7 %, and the 7.5 % road from A1 is only about 4.3°.

Notice that the percent grows much faster than the angle. That is why "a 100 % slope" sounds vertical but is only 45°.

## 8. Functions inside loops

Here is where functions pay off. Once a function works for one value, a loop can apply it to a whole list.

There are two ways to collect the results. You can start an empty list and `append` each result inside a `for` loop. Or you can write a list comprehension, as in A3: `[f(x) for x in values]`.

The slope angles below are made-up values for practice.

````python
slopes_deg = [5, 10, 20, 30, 45]   # made-up survey values

slopes_percent = []
for slope_deg in slopes_deg:
    slopes_percent.append(round(degrees_to_percent(slope_deg), 1))
print(slopes_percent)
````

````python
heights_m = [8849, 8586, 8516]
heights_ft = [round(m_to_ft(height_m)) for height_m in heights_m]
print(heights_ft)
````

The loop called `degrees_to_percent` five times: `[8.7, 17.6, 36.4, 57.7, 100.0]`. The list comprehension did the same job for feet in one line: `[29032, 28169, 27940]`.

## 9. When it breaks: a missing argument

A very common mistake with functions is to forget an argument. The cell below calls `gradient_percent` with only the rise. Run it.

````python
gradient_percent(150)
````

Read from the bottom up:

- `TypeError` is the type of error. Here it means the function was called the wrong way.
- `gradient_percent() missing 1 required positional argument: 'run_m'` says exactly what is missing. A **positional argument** is one matched by its position.

`digits` was not mentioned, because it has a default. Only parameters without a default are **required**.

The opposite mistake gives a similar message: `m_to_ft(8849, 8586)` would say `m_to_ft() takes 1 positional argument but 2 were given`.

def roof_water_l(rain_mm, roof_area_m2=50):
    """Return the litres of rain collected by a roof."""
    return rain_mm * roof_area_m2

total_l = 0
for rain_mm in [16, 23, 28, 54]:
    water_l = roof_water_l(rain_mm)
    print(f"{rain_mm} mm -> {water_l:,} litres")
    total_l += water_l
print(f"Total: {total_l:,} litres")

## Recap

- `def name(parameters):` creates a **function**; the indented body runs only when you **call** it.
- Arguments are matched to parameters by position, unless you name them with **keyword arguments**.
- `return` hands a value back so you can store it and calculate with it; a function without `return` gives `None`.
- Default values, docstrings and small functions that call each other make your tools clear and reusable.
- `import math` gives `math.tan`, `math.atan`, `math.radians`, `math.degrees` and `math.pi`; angles are in radians.

### Words you learned

| Word | Meaning |
| --- | --- |
| function | A named block of code that runs when you call it |
| `def` | The word that starts a function definition |
| body | The indented lines inside a function |
| call | Running a function by writing its name and brackets, like `m_to_ft(8849)` |
| parameter | A name in the `def` line that receives a value, like `height_m` |
| argument | The value you pass in when you call a function, like `8849` |
| `return` | Hands a value back to the caller and ends the function |
| `None` | Python's value for "nothing"; what a function without `return` gives back |
| default value | A value a parameter uses when the caller leaves it out, like `digits=1` |
| keyword argument | An argument passed with its name, like `digits=3` |
| positional argument | An argument matched to a parameter by its position |
| docstring | A description in triple quotes at the top of a function's body |
| module | A toolbox of extra functions that you load with `import`, like `math` |
| radian | The unit `math` uses for angles; 360° is 2π radians |
| slope angle | Steepness as an angle in degrees from the horizontal |
