## 1. Four kinds of values

Every value in Python has a **type**. The type decides what you can do with the value. These four types cover most everyday work:

| Type | Name in Python | Examples |
| --- | --- | --- |
| whole number | `int` (integer) | `8849`, `77`, `-3` |
| decimal number | `float` | `8.849`, `27.9881`, `3.28084` |
| text | `str` (string) | `"Everest"`, `"Koshi"` |
| yes or no | `bool` (Boolean) | `True`, `False` |

`True` and `False` start with a capital letter and have no quotation marks. The function `type()` tells you the type of any value.

````python
everest_m = 8849
everest_km = 8.849
peak_name = "Everest"
is_in_nepal = True

print(type(everest_m))
print(type(everest_km))
print(type(peak_name))
print(type(is_in_nepal))
````

Python writes `<class 'int'>`. Here **class** is another word for type.

**Predict:** `"8849"` has quotation marks around it. Is it a number or text? Write down your guess, then run the cell.

````python
print(type("8849"))
print(type(8849.0))
````

`"8849"` is text, because of the quotation marks. To Python it is four characters, not a height. `8849.0` is a `float`, because it has a decimal point, even though the decimal part is zero.

## 2. Converting between types

Data often arrives as text, for example when it is read from a file (Lesson A7). Before you can calculate with it, you must convert it to a number.

| Function | Converts to | Example | Result |
| --- | --- | --- | --- |
| `int()` | whole number | `int("8849")` | `8849` |
| `float()` | decimal number | `float("27.9881")` | `27.9881` |
| `str()` | text | `str(8849)` | `"8849"` |

`int()` on a decimal number cuts off the decimal part; it does **not** round. `int(8.9)` gives `8`. Use `round()` when you want rounding.

The text must look like a plain number. `int("8,849")` fails because of the comma; Lesson A6 shows how to handle that.

````python
height_text = "8849"           # text, as if read from a file
height_m = int(height_text)    # now a whole number
print(height_m + 1)
print(float("27.9881"))
print(int(8.9))
print(round(8.9))
````

````python
str(8849)
````

The second cell shows `'8849'` with quotation marks. That is Python's way of telling you the result is text, not a number.

## 3. Whole parts and leftovers: `//` and `%`

Dividing with `/` always gives a decimal answer. Sometimes you want whole units and what is left over instead, like "8 km and 849 m".

- `//` is **floor division**: how many *whole* times one number fits into another.
- `%` is the **remainder** (also called *modulo*): what is left over after that.

They work as a pair. `135 // 60` is `2` and `135 % 60` is `15`: 135 minutes is 2 hours and 15 minutes.

**Predict:** before you run the cell, write down what `8849 // 1000` and `8849 % 1000` will give.

````python
everest_m = 8849
print(everest_m / 1000)
print(everest_m // 1000)    # whole kilometres
print(everest_m % 1000)     # metres left over
````

Everest is 8.849 km high: that is 8 whole kilometres and 849 m left over.

## 4. Comparisons: True or False

A **comparison** asks Python a yes-or-no question. The answer is a `bool`: `True` or `False`.

| Symbol | Question | Example | Result |
| --- | --- | --- | --- |
| `==` | equal? | `8849 == 8849` | `True` |
| `!=` | not equal? | `8849 != 8586` | `True` |
| `>` | greater than? | `8586 > 8849` | `False` |
| `<` | less than? | `822 < 1400` | `True` |
| `>=` | greater than or equal? | `8000 >= 8000` | `True` |
| `<=` | less than or equal? | `8091 <= 8000` | `False` |

Do not mix up `=` and `==`. A single `=` *stores* a value; a double `==` *asks* whether two values are equal.

Text can be compared too. Comparisons of text are case-sensitive, like variable names.

````python
everest_m = 8849
kanchenjunga_m = 8586

print(everest_m > kanchenjunga_m)
print(everest_m == 8849)
print(kanchenjunga_m >= 9000)
print("Everest" == "everest")
````

````python
is_higher = everest_m > kanchenjunga_m
print(is_higher, type(is_higher))
````

The first cell prints `True`, `True`, `False`, `False`. The last one is `False` because a capital `E` and a small `e` are different characters.

The second cell shows that you can store the answer of a comparison in a variable, like any other value.

## 5. Joining text: `+`, `*` and `len()`

The `+` sign joins two strings into one. This is called **concatenation**. Python adds no spaces for you, so put them inside the quotation marks yourself.

The `*` sign repeats a string: `"-" * 5` gives `"-----"`. It is handy for drawing lines under titles.

`len()` counts the characters in a string. Spaces and punctuation count too.

You can only join text to text. To join a number, convert it first with `str()`.

````python
peak = "Everest"
province = "Koshi"
label = peak + ", " + province + " Province"
print(label)
print("-" * 20)
print(len(peak))
print(len(label))
````

````python
print("Everest is " + str(8849) + " m high")
````

`"Everest, Koshi Province"` has 23 characters: 7 for `Everest`, 2 for the comma and space, 5 for `Koshi` and 9 for `" Province"`.

## 6. String methods: cleaning and cutting text

A **method** is a function that belongs to a value. You write it after the value with a dot: `name.upper()`. The brackets are needed, even when they are empty.

| Method | What it does | Example | Result |
| --- | --- | --- | --- |
| `.upper()` | capital letters | `"koshi".upper()` | `"KOSHI"` |
| `.lower()` | small letters | `"KOSHI".lower()` | `"koshi"` |
| `.title()` | capital first letter of each word | `"cho oyu".title()` | `"Cho Oyu"` |
| `.replace(old, new)` | swap one piece of text for another | `"Cho-Oyu".replace("-", " ")` | `"Cho Oyu"` |
| `.strip()` | remove spaces at both ends | `"  Lukla ".strip()` | `"Lukla"` |
| `.split(sep)` | cut into pieces at `sep` | `"a,b".split(",")` | `['a', 'b']` |

A method never changes the original string. It gives back a new one, so store it if you want to keep it.

`.split()` with nothing in the brackets cuts at spaces. Its result, in square brackets, is a **list**: several values kept in order. Lesson A3 is all about lists.

````python
raw_name = "  dhaulagiri i  "      # typed carelessly
print(raw_name.strip())
print(raw_name.strip().title())
print("koshi".upper())
print("Cho-Oyu".replace("-", " "))
````

````python
coordinates = "27.9881,86.9250"
print(coordinates.split(","))
print("Koshi Bagmati Gandaki".split())
````

In `raw_name.strip().title()`, Python works from left to right. `.strip()` removes the spaces, then `.title()` capitalises the result. Using several methods in a row like this is called **chaining**.

## 7. Picking one character: indexing

Each character in a string has a position number, called its **index**. Python counts from **0**, not from 1.

| Character | `M` | `a` | `k` | `a` | `l` | `u` |
| --- | --- | --- | --- | --- | --- | --- |
| Index | 0 | 1 | 2 | 3 | 4 | 5 |

To pick one character, write the index in square brackets after the string: `peak[0]` is the first character. The last index is always the length minus one.

**Predict:** in the cell below, `peak` holds `"Makalu"`. What will `peak[1]` show?

````python
peak = "Makalu"
print(peak[0])
print(peak[1])
print(peak[5])
print(peak[len(peak) - 1])    # the last character, whatever the length
````

`peak[1]` is `a`, the *second* letter, because counting starts at 0. `"Makalu"` has 6 characters, so the last one is at index 5. Lesson A3 uses the same square brackets on lists.

## 8. f-strings: clear sentences with numbers and units

Joining with `+` and `str()` works, but it is slow to type and mistakes creep in. An **f-string** is better. Put the letter `f` before the opening quotation mark, then write any value or calculation inside curly brackets `{ }`.

After the value you can add a colon and a **format code**:

| Code | Meaning | Example | Result |
| --- | --- | --- | --- |
| `{x:,}` | commas between thousands | `f"{8849:,}"` | `8,849` |
| `{x:.1f}` | one decimal place | `f"{8.849:.1f}"` | `8.8` |
| `{x:,.0f}` | commas and no decimals | `f"{29031.95:,.0f}"` | `29,032` |

Always write the unit after the number, like `m`, `ft` or `%`. A number without a unit is a guess for the reader.

````python
peak = "Everest"
height_m = 8849
height_ft = height_m * 3.28084

print(f"{peak} is {height_m:,} m high.")
print(f"That is about {height_ft:,.0f} ft.")
print(f"In kilometres: {height_m / 1000:.1f} km")
````

````python
population = 29164578    # Nepal, 2021 census
print(f"Population of Nepal in 2021: {population:,}")
````

The f-string turns numbers into text for you, so `str()` is not needed. The format codes only change how the number is *shown*: `height_ft` itself still holds 29,031.95… with all its decimals.

## 9. When it breaks: mixing text and numbers

This is the most common mistake with text. The cell tries to join a string and a number with `+`. Run it and read the message.

````python
height_m = 8849
print("Everest is " + height_m + " m high")
````

Read the message from the bottom up:

- `TypeError` is the type of error: an operation got a value of the wrong **type**.
- `can only concatenate str (not "int") to str` is the reason: `+` can join text to text, but `height_m` is an `int`.

The arrow points to the line with the problem. There are two ways to fix it:

```python
print("Everest is " + str(height_m) + " m high")
print(f"Everest is {height_m:,} m high")
```

The f-string version is shorter and adds the comma too. When you see `TypeError`, check the type of each value with `type()`.

## Recap

- Every value has a **type**: `int`, `float`, `str` or `bool`. `type()` tells you which.
- `int()`, `float()` and `str()` convert between types. `int()` cuts off decimals; `round()` rounds.
- `//` gives whole parts and `%` gives the leftover. Comparisons like `>` and `==` give `True` or `False`.
- Strings can be joined (`+`), repeated (`*`), measured (`len()`), cleaned with methods and indexed from 0.
- f-strings like `f"{height_m:,} m"` are the clearest way to write numbers with their units.

### Words you learned

| Word | Meaning |
| --- | --- |
| type | The kind of a value, such as whole number or text |
| `int` | A whole number, like `8849` |
| `float` | A decimal number, like `8.849` |
| `str` (string) | Text inside quotation marks |
| `bool` (Boolean) | One of the two values `True` or `False` |
| converting | Turning a value into another type, with `int()`, `float()` or `str()` |
| floor division `//` | Division that keeps only the whole part |
| remainder `%` | What is left over after floor division (also called modulo) |
| comparison | A yes-or-no question like `a > b`, answered with `True` or `False` |
| concatenation | Joining strings with `+` |
| method | A function that belongs to a value, written after a dot: `name.upper()` |
| chaining | Using several methods in a row: `name.strip().title()` |
| list | Several values in square brackets, kept in order (Lesson A3) |
| index | The position of a character, counted from 0 |
| f-string | A string starting with `f` that fills in values written in `{ }` |
| format code | Instructions after a colon in `{ }`, like `:,` or `:.1f` |
| `TypeError` | An error raised when a value has the wrong type for an operation |
