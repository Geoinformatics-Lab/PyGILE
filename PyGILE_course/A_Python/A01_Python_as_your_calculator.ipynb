## 2. Python as a calculator

Python can do arithmetic, like a calculator. Mount Everest is **8,849 m** high, and Kathmandu sits at about **1,400 m**.

| Symbol | Meaning | Example | Result |
| --- | --- | --- | --- |
| `+` | add | `8849 + 1` | `8850` |

| `**` | power | `10 ** 3` | `1000` |

Write `8849`, never `8,849`: in Python a comma separates two different things.

## 3. Order of operations and brackets

Python follows the same order as mathematics:

1. brackets `( )` first,
2. then powers `**`,
3. then `*` and `/`,
4. then `+` and `-`.

When you are not sure, add brackets. They never hurt, and they make your intention clear to anyone reading the code.

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
````

The first line multiplies before it adds (14). In the second, the brackets force the addition first (20).

## 4. Variables: giving a value a name

Typing `8849` again and again is slow, and mistakes creep in. Instead, you can store a value under a name. A value stored under a name is a **variable**.

Read `everest = 8849` as **"store 8849 under the name `everest`"**. The `=` sign does not mean "equals" as in mathematics. It means: work out the right-hand side, then store it in the name on the left.

Storing a value shows nothing. That is normal.

````python
everest = 8849
kanchenjunga = 8586

everest - kanchenjunga
````

````python
difference = everest - kanchenjunga
print(difference)
````

## 6. Good names and comments

Variable names follow a few rules:

| Allowed | Not allowed | Why |
| --- | --- | --- |
| `everest_height` | `everest height` | No spaces. Use `_` (underscore). |
| `peak_2` | `2nd_peak` | A name cannot start with a number. |
| `rainfall_mm` | `rainfall-mm` | `-` means minus. |

Python is **case-sensitive**: `everest` and `Everest` are two different names. Use lowercase.

Good names say what they hold and include the unit, like `distance_m` or `area_km2`.

A **comment** starts with `#`. Python ignores everything after it on that line. Comments are notes for people, including you next month.

````python
lukla_m = 2860       # Lukla airstrip, start of the Everest trek
base_camp_m = 5364   # Everest Base Camp

# total height gained on the trek
print(base_camp_m - lukla_m)
````

## 7. A real calculation: steep roads and other units

Engineers describe a road's steepness as a **gradient** in percent: how many metres it climbs for every 100 m travelled horizontally.

gradient (%) = rise ÷ run × 100

Both the rise and the run must be in the same unit. Here a road climbs 150 m over a horizontal distance of 2 km.

````python
rise_m = 150
run_m = 2000          # 2 km written in metres

gradient_percent = rise_m / run_m * 100
print(gradient_percent)
````

````python
metres_to_feet = 3.28084
everest_ft = everest * metres_to_feet
print(everest_ft)
print(round(everest_ft))
print(round(everest_ft, 1))
````

The road has a gradient of 7.5%.

Everest is about 29,032 ft. The last digits of the long number (`...999998`) look odd: computers store decimals in binary, so tiny differences like this are normal and harmless.

`round(value)` rounds to a whole number; `round(value, 1)` keeps one decimal place. `round` is a **function**: a named tool that takes values in brackets and gives back a result. `print` is a function too. In Lesson A5 you will write your own.
