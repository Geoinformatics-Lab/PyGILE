## 1. Lists: many values under one name

You met a list in Lesson A2: `.split()` gives one back. Now you will make your own.

## 2. Picking one item: indexing

Items in a list are numbered from **0**, exactly like characters in a string (Lesson A2). `peaks[0]` is the first item and `peaks[7]` the eighth.

You can also count from the end with a **negative index**. `peaks[-1]` is the last item and `peaks[-2]` the second-to-last. This works for strings too: `"Makalu"[-1]` is `"u"`.

| Item | Everest | Kanchenjunga | … | Manaslu | Annapurna I |
| --- | --- | --- | --- | --- | --- |
| Index | 0 | 1 | … | 6 | 7 |
| Negative index | -8 | -7 | … | -2 | -1 |

**Predict:** what will `peaks[3]` show? Count carefully from 0.

````python
print(peaks[0])
print(peaks[3])
print(heights_m[3])
print(peaks[-1])
print(heights_m[-1])
````

`peaks[3]` is Makalu, the *fourth* peak. Because the two lists line up, `heights_m[3]` is Makalu's height, 8,485 m. The same index always picks matching items from both lists.

## 3. Taking a piece: slicing

A **slice** takes several items at once. Write `list[start:stop]`. You get the items from `start` up to, but **not including**, `stop`.

- `peaks[0:3]` gives items 0, 1 and 2: three items.
- Leave out `start` to begin at the start: `peaks[:3]`.
- Leave out `stop` to go to the end: `peaks[5:]`.
- Negative numbers count from the end: `peaks[-3:]` is the last three.

The number of items in a slice is `stop - start`. Slices work on strings too, which is very useful for codes and dates.

````python
print(peaks[0:3])
print(peaks[:3])
print(heights_m[5:])
print(peaks[-3:])
print("LS017"[2:])
````

A slice of a list is a new, shorter list. A slice of a string is a new, shorter string: `"LS017"[2:]` is `"017"`, the number part of a landslide code.

## 4. Changing a list

Unlike a string, a list can be changed after you make it. These tools change the list itself:

| Code | What it does |
| --- | --- |
| `trek.append(x)` | adds `x` at the end |
| `trek.insert(i, x)` | puts `x` at index `i` and moves the rest along |
| `trek.remove(x)` | removes the first item equal to `x` |
| `trek[i] = x` | replaces the item at index `i` with `x` |

`.append()`, `.insert()` and `.remove()` are methods, like the string methods in Lesson A2. They change the list and give back nothing, so do not write `trek = trek.append(x)`.

````python
trek = ["Lukla", "Namche Bazaar"]
trek.append("Everest Base Camp")
print(trek)
trek.insert(0, "Kathmandu")
print(trek)
trek.remove("Kathmandu")
print(trek)
````

````python
ascent_years = [1953, 1955, 1965]    # Lhotse typed wrongly as 1965
ascent_years[2] = 1956                # replace the item at index 2
print(ascent_years)

gandaki_peaks = ["Dhaulagiri I", "Annapurna I"]
````

Running the first cell twice adds `Everest Base Camp` twice, because the list keeps its changes. If a list looks wrong, restart the kernel and run all cells again.

## 6. for loops: repeating for every item

A **loop** repeats some code. A `for` loop repeats it once for every item in a list:

```python
for peak in peaks:
    print(peak)
```

Read it as: *for each item in `peaks`, call it `peak`, then run the indented lines.* The variable `peak` holds a different item each time round. Each time round is called an **iteration**.

The first line ends with a colon `:`. The lines that repeat are **indented** by four spaces; together they are called the loop's **block**. JupyterLab indents for you after a colon. The first line that is not indented runs once, after the loop has finished.

A good habit: name the list in the plural and the loop variable in the singular, like `peaks` and `peak`.

**Predict:** how many lines will the first cell below print?

````python
for peak in peaks:
    print(peak)
print("Done")
````

````python
for height_m in heights_m:
    height_ft = height_m * 3.28084
    print(f"{height_m:,} m = {height_ft:,.0f} ft")
````

The first cell prints 9 lines: 8 names from the loop, then `Done` once, because that line is not indented.

In the second cell, both indented lines run for every height. The loop does the same work eight times, and you wrote it only once.

## 7. Counting with `range()` and `enumerate()`

`range()` makes a sequence of whole numbers to loop over:

| Code | Numbers |
| --- | --- |
| `range(4)` | 0, 1, 2, 3 |
| `range(2, 5)` | 2, 3, 4 |
| `range(3000, 5501, 500)` | 3000, 3500, 4000, 4500, 5000, 5500 |

Like a slice, `range` stops *before* the stop number. The third number is the step.

`range(len(peaks))` gives every index of `peaks`: 0 to 7. Loop over it when you need the same index in two lists.

`enumerate()` gives you the index and the item together. Write two loop variables, separated by a comma. With `start=1` it counts from 1, which suits rankings.

````python
for altitude_m in range(3000, 5501, 500):
    print(altitude_m)
````

````python
for i in range(len(peaks)):
    print(peaks[i], heights_m[i])
````

````python
for rank, peak in enumerate(peaks, start=1):
    print(f"{rank}. {peak}")
````

In the second cell, `i` is 0, then 1, up to 7, and picks matching items from both lists. In the third, `rank` counts 1 to 8 while `peak` takes each name.

## 8. Building results in a loop

Loops often build up an answer, one item at a time. The pattern has three parts:

1. **Before** the loop, create a starting value: `0` for a total, or `[]` for an empty list.
2. **Inside** the loop, update it.
3. **After** the loop, use the result.

`total += x` is short for `total = total + x`. A total that grows like this is called a **running total**.

To build a new list, start with `[]` and `.append()` inside the loop. A **list comprehension** does the same in one line: `[height_m / 1000 for height_m in heights_m]`. Read it as *"height divided by 1,000, for each height in the list"*.

The cell below follows a trek from Lukla (2,860 m) with made-up daily height gains. **Predict:** where does the trekker end up after seven days?

````python
daily_gain_m = [580, 0, 400, 450, 0, 520, 554]    # made-up values for practice; 0 = rest day

altitude_m = 2860              # start at Lukla
for gain_m in daily_gain_m:
    altitude_m += gain_m       # same as altitude_m = altitude_m + gain_m
    print(altitude_m)
````

````python
heights_ft = []                           # 1. start with an empty list
for height_m in heights_m:
    heights_ft.append(round(height_m * 3.28084))    # 2. add one item each time
print(heights_ft)                         # 3. use the result
````

````python
heights_km = [height_m / 1000 for height_m in heights_m]
print(heights_km)
````

The trekker reaches 3,440 m (Namche Bazaar) on day 1 and 5,364 m (Everest Base Camp) on day 7. On rest days the altitude stays the same.

The last two cells both build a new list from `heights_m`. The comprehension is shorter; the loop is easier to extend with more lines.

## Recap

- A **list** keeps values in order: `heights_m = [8849, 8586, ...]`. `len()` counts its items.
- Indexes start at 0; negative indexes count from the end. A slice `[start:stop]` stops before `stop`.
- `.append()`, `.insert()`, `.remove()` and `list[i] = x` change a list; `sorted()` gives a new one.
- `sum()`, `min()`, `max()` and `in` summarise a list without a loop.
- A `for` loop repeats its indented block for every item. `range()` and `enumerate()` give you counting numbers, and `+=` or `.append()` build results.

### Words you learned

| Word | Meaning |
| --- | --- |
| list | Several values in square brackets, kept in order |
| item | One value in a list |
| negative index | A position counted from the end: `-1` is the last item |
| slice | A piece of a list or string: `peaks[1:4]` |
| `append`, `insert`, `remove` | List methods that add or remove items |
| `sorted()` | Gives a new list in order, smallest or A first |
| `in` | Asks whether a value is an item of a list; gives `True` or `False` |
| loop | Code that repeats |
| `for` loop | A loop that runs once for every item in a list or range |
| iteration | One time round a loop |
| block | The indented lines that belong to a loop |
| indentation | Spaces at the start of a line; four spaces mark a block |
| `range()` | Gives a sequence of whole numbers, stopping before the stop value |
| `enumerate()` | Gives the index and the item together in a loop |
| running total | A total that grows inside a loop with `+=` |
| list comprehension | A one-line way to build a list: `[x / 1000 for x in values]` |
| `IndexError` | An error raised when an index is outside the list |
