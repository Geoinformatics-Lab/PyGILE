## 1. The anatomy of a traceback

When Python stops at an error, it prints a **traceback**: the route it followed to reach the problem.

A traceback has two parts:

- **The last line** gives the error **type** (such as `TypeError`) and a **message** explaining it. Always read this first.
- **The levels above it** show the code that was running. Each level names a place, such as `Cell In[5], line 8`, and an arrow `---->` points at the line.

When a function calls another function, there are several levels. The top level is the line you ran. The bottom level is where Python finally broke. The *cause* is often higher up, in the value that was passed in.

**Predict:** the cell below calls a function with the height written in quotes. Which line do you think will fail?

````python
def m_to_ft(length_m):
    return length_m * 3.28084

def describe_in_feet(name, height_m):
    height_ft = round(m_to_ft(height_m))
    return f"{name}: {height_ft:,} ft"

describe_in_feet("Everest", "8849")
````

Read it from the bottom up:

1. **Last line:** `TypeError: can't multiply sequence by non-int of type 'float'`. A **sequence** here is the string `"8849"`. Python cannot multiply text by 3.28084.
2. **Bottom level (line 2):** the multiplication inside `m_to_ft`. This is where Python broke.
3. **Middle level (line 5):** `describe_in_feet` called `m_to_ft`.
4. **Top level (line 8):** the line you ran. Here is the real cause: `"8849"` is text, because of the quotation marks.

Python broke in `m_to_ft`, but `m_to_ft` is fine. The mistake was one level up. This is very common.

Both functions were still defined, because their `def` lines ran before the error.

## 2. SyntaxError and IndentationError: Python cannot read the code

Some errors happen before a single line runs. Python first reads the whole cell. If the grammar is broken, it refuses to start and raises a **SyntaxError**.

A SyntaxError has no levels. It shows the line and puts a caret `^` under the place where Python got confused. The real mistake is often right before the caret: a missing colon, bracket or quotation mark.

An **IndentationError** is a kind of SyntaxError. It means the spaces at the start of a line are wrong. The body of a `for`, `if` or `def` must be indented, usually by 4 spaces.

````python
peaks = ["Everest", "Lhotse", "Makalu"]
for peak in peaks
    print(peak)
````

````python
def m_to_km(length_m):
return length_m / 1000
````

The first message is `SyntaxError: expected ':'`: the `for` line needs a colon at the end.

The second is `IndentationError: expected an indented block after function definition on line 1`. Python even tells you which line started the block. Notice that nothing ran: `peaks` was never created, because the whole cell was refused.

Two more SyntaxError messages you will meet: `unterminated string literal` (a quotation mark is missing) and `'(' was never closed` (a bracket is missing).

## 3. NameError, TypeError and AttributeError: wrong name or wrong kind

These three errors are about **what** you asked Python to use.

- **NameError:** Python does not know a name. It may be misspelt, have the wrong capital letters, or the cell that creates it was not run yet.
- **TypeError:** the operation does not suit the type of value, like adding text and a number. Calling a function with the wrong number of arguments is also a TypeError, as you saw in A5.
- **AttributeError:** you asked a value for a method it does not have. For example, `.upper()` belongs to strings, not to lists.

The message names the types involved. Read it carefully: it usually tells you the fix.

- `AttributeError: 'list' object has no attribute 'upper'`. A list has no `.upper()`; each string inside it does.

- **ZeroDivisionError:** you divided by zero. In real data this often means a missing value was stored as 0.

## 6. A debugging routine

Some bugs give no error at all, only a wrong answer. These are harder to find. A **bug** is any mistake in code; **debugging** is finding and fixing it. Use the same calm routine every time:

1. **Read the traceback bottom-up.** Find the last line, then the level that points at your own code.
2. **Check the types** with `type()`. Many errors come from a number stored as text.
3. **Print intermediate values**, especially inside loops, to see where the numbers go wrong.
4. **Test a small piece** on its own, with a value whose answer you already know.
5. **Search the message.** Copy the last line, remove your own names and values, and add the word *python*. For example: `python TypeError can only concatenate str (not "int") to str`. Read more than one answer, and prefer ones that explain *why*.

The function below should add up a year of rainfall. The values are Station A from the course file `practice_rainfall.csv` (made-up practice data). **Predict:** does it give the right total?

````python
def total_rainfall_mm(monthly_mm):
    for value_mm in monthly_mm:
        total_mm = 0
        total_mm += value_mm
    return total_mm

station_a_mm = [35, 59, 78, 151, 206, 536, 1045, 952, 592, 198, 42, 42]
print(total_rainfall_mm(station_a_mm))
````

The answer is 42, the value of the last month. No error, but clearly wrong: a year of monsoon rain is not 42 mm. Let us add a debugging `print` inside the loop and test a short list.

````python
def total_rainfall_mm(monthly_mm):
    for value_mm in monthly_mm:
        total_mm = 0
        total_mm += value_mm
        print(value_mm, total_mm)    # debugging print
    return total_mm

total_rainfall_mm([35, 59, 78])
````

The running total never grows: it is 35, then 59, then 78. The line `total_mm = 0` is inside the loop, so the total is reset to zero every month. It belongs *before* the loop.

## 7. `try` and `except`: planning for problems you expect

Some errors are not mistakes in your code. They are problems you *expect* in the data, such as a rainfall reading of `"n/a"` or an empty cell.

`try` and `except` let you plan for them:

```
try:
    lines that might fail
except ValueError:
    what to do instead
```

Python runs the `try` block. If a `ValueError` happens there, it jumps to the `except` block instead of stopping.

Always name the error you expect, such as `except ValueError:`. An `except` without a name catches every error, including your own typos, and hides real bugs.

````python
readings = ["35", "59", "n/a", "151", "", "536"]   # made-up messy rainfall text, in mm

clean_mm = []
for text in readings:
    try:
        clean_mm.append(float(text))
    except ValueError:
        print(f"Skipped '{text}'")

print(clean_mm)
````

`float("n/a")` and `float("")` raised `ValueError`, so the `except` block printed a note and the loop went on. The four good readings were kept: `[35.0, 59.0, 151.0, 536.0]`.

- The level where Python broke is not always the cause; check the values passed in from the levels above.
- SyntaxError and IndentationError stop the whole cell before it runs; most other errors happen while it runs.
- Debugging routine: read the traceback, check types with `type()`, print intermediate values, test small pieces, search the message.
- `try`/`except SomeError:` handles problems you expect in the data. Always name the error you expect.
