## 1. Anatomy of a figure

**Matplotlib** is a library for drawing charts. Its main tools are loaded with `import matplotlib.pyplot as plt`.

Every chart has two parts. The **figure** is the whole image, like a sheet of paper. An **axes** is one plotting area on that sheet, with an x-axis (across) and a y-axis (up). One figure can hold several axes.

`fig, ax = plt.subplots()` makes a figure with one axes and stores them in two names at once, as `row, col = ...` did in Lesson A8. `plt.subplots(figsize=(8, 3))` sets the size in inches: width, then height.

You then draw on the axes with its methods: `ax.plot(x_values, y_values)` draws a line. `ax.set_title()`, `ax.set_xlabel()` and `ax.set_ylabel()` add text. `plt.show()` displays the finished figure below the cell.

The first cell below loads everything this lesson needs. **Predict:** in the second cell, will the line go up steadily, or bend?

````python
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path

DATA = Path("..") / "data"
OUT = Path("..") / "outputs"
OUT.mkdir(exist_ok=True)

peaks = pd.read_csv(DATA / "nepal_peaks.csv")
rain = pd.read_csv(DATA / "practice_rainfall.csv")
slides = pd.read_csv(DATA / "practice_landslides.csv")
````

````python
# a trek from Lukla to Everest Base Camp (the day numbers are a made-up itinerary)
trek_day = [1, 3, 8]
trek_alt_m = [2860, 3440, 5364]   # Lukla, Namche Bazaar, Everest Base Camp

fig, ax = plt.subplots()
ax.plot(trek_day, trek_alt_m, marker="o")
ax.set_title("Trek from Lukla to Everest Base Camp")
ax.set_xlabel("Day of trek")
ax.set_ylabel("Elevation (m)")
plt.show()
````

The line bends upwards: the climb from Namche Bazaar to Base Camp is steeper per day than the climb from Lukla to Namche. `marker="o"` puts a dot on each data point, so you can see where the real values are.

Look at the parts. The title is at the top, and the axis labels carry their **units** in brackets. Matplotlib chose the **tick labels**, the numbers along each axis, for you.

## 2. A line chart: monthly rainfall at one station

A **line chart** shows how a value changes along an ordered sequence, such as days or months. The line joins neighbours, so the order must mean something.

To plot one station's year, take its row from the rainfall table, as in Lesson A9: make the station names the index, then use `.loc` with the label slice `"jan_mm":"dec_mm"`. The result is a Series of twelve values.

The x values can be text. A list of twelve month names puts one label under each point.

````python
stations = rain.set_index("station")
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
station_a_mm = stations.loc["Station A", "jan_mm":"dec_mm"]

fig, ax = plt.subplots()
ax.plot(months, station_a_mm, marker="o")
ax.set_title("Monthly rainfall at Station A, Kaski (practice data)")
ax.set_xlabel("Month")
ax.set_ylabel("Rainfall (mm)")
plt.show()
````

The monsoon stands out at once: rainfall climbs from May, peaks in July and falls away by October. The title says the data are practice data, so no reader mistakes them for measurements.

## 3. Several lines and a legend

To compare stations, draw several lines on the **same axes**: call `ax.plot` once per line. Matplotlib gives each line a new colour.

Give each line a name with `label="Station A"`, then call `ax.legend()`. The **legend** is the key that says which colour is which line.

A loop over station names saves typing. Keep it to a few lines, about four at most: more lines turn the chart into a tangle that nobody can read.

**Predict:** which of the three stations will have the highest point in July?

````python
fig, ax = plt.subplots()
for name in ["Station A", "Station B", "Station C"]:
    monthly_mm = stations.loc[name, "jan_mm":"dec_mm"]
    ax.plot(months, monthly_mm, marker="o", label=name)

ax.set_title("Monthly rainfall at three stations (practice data)")
ax.set_xlabel("Month")
ax.set_ylabel("Rainfall (mm)")
ax.legend()
plt.show()
````

Station A towers over the others in July. Now that all three share one y-axis, Station C's monsoon, which looked large on its own chart, is a flat line near the bottom.

## 4. Bar charts: comparing the peaks

A **bar chart** compares amounts across named things, such as peaks or districts. The length of each bar shows the amount.

`ax.bar(names, values)` draws vertical bars. `ax.barh(names, values)` draws **horizontal** bars, which leave room for long names like "Kanchenjunga".

Sort the table first, so the bars form a ranking that the eye can read. `barh` draws the first row at the **bottom**, so sort from smallest to largest to put the tallest bar at the top.

A bar's length is its message, so bars must **start at zero**. Matplotlib does this for you unless you change it.

````python
sorted_peaks = peaks.sort_values("height_m")

fig, ax = plt.subplots()
ax.barh(sorted_peaks["name"], sorted_peaks["height_m"])
ax.set_title("Nepal's eight peaks above 8,000 m")
ax.set_xlabel("Height (m)")
plt.show()
````

Everest is at the top and Annapurna I at the bottom. The bars look almost equal, and that is the honest truth: the difference between the tallest and the shortest is small compared with their full height.

## 5. Scatter plots: two numbers for each record

A **scatter plot** draws one dot per record, using one number for the x position and another for the y position. It helps you see whether two quantities move together.

`ax.scatter(x_values, y_values)` draws the dots. Unlike `ax.plot`, it does not join them with a line, because the records have no order.

`ax.text(x, y, "words")` writes text at a position on the axes. In a loop it can label each dot.

**Predict:** Station C is the highest station and the driest. Where on the chart will its dot be?

````python
rain["annual_mm"] = rain.loc[:, "jan_mm":"dec_mm"].sum(axis=1)

fig, ax = plt.subplots()
ax.scatter(rain["elevation_m"], rain["annual_mm"])
for i in range(len(rain)):
    row = rain.iloc[i]
    ax.text(row["elevation_m"] + 40, row["annual_mm"], row["station"])

ax.set_title("Annual rainfall and elevation (practice data)")
ax.set_xlabel("Station elevation (m)")
ax.set_ylabel("Annual rainfall (mm)")
plt.show()
````

Station C sits at the bottom right: high and dry. But there is no simple pattern. Station D, the lowest station, is drier than Station F at 1,600 m, and Station A is by far the wettest.

Six made-up stations are far too few to show how rainfall depends on elevation. A scatter plot shows what the data contain; it cannot prove a rule.

`+ 40` moves each label 40 m to the right of its dot, so the text does not cover it.

## 6. Several plots side by side

`plt.subplots(1, 2)` makes a figure with **1 row and 2 columns** of axes. Small charts placed side by side are called **subplots**.

Unpack them with `fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))`. Each axes gets its own plot, title and labels.

Add `sharey=True` to give both plots the **same y-axis**. Then equal heights mean equal values, and the comparison is fair.

`fig.suptitle("...")` adds one title over the whole figure. `fig.tight_layout()` moves the plots apart so their labels do not overlap.

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4), sharey=True)

ax1.plot(months, stations.loc["Station A", "jan_mm":"dec_mm"], marker="o")
ax1.set_title("Station A, Kaski")
ax1.set_xlabel("Month")
ax1.set_ylabel("Rainfall (mm)")

ax2.plot(months, stations.loc["Station C", "jan_mm":"dec_mm"], marker="o")
ax2.set_title("Station C, Mustang")
ax2.set_xlabel("Month")

fig.suptitle("Monthly rainfall (practice data)")
fig.tight_layout()
plt.show()
````

With a shared y-axis, the contrast is honest: Kaski's monsoon fills the left panel, while Mustang barely leaves the bottom of the right one. The y-axis numbers are printed once, on the left.

## 7. Saving a figure

`fig.savefig(path)` saves the figure as an image file. The ending of the file name picks the format: `.png` for reports and slides, `.pdf` or `.svg` for drawings that stay sharp at any size.

Two settings matter:

- `dpi=300` sets the resolution: 300 **dots per inch** is good enough for print.
- `bbox_inches="tight"` trims empty margins and keeps labels from being cut off.

Save **before** `plt.show()`. In some programs, showing a figure clears it, and you would save an empty image.

````python
fig, ax = plt.subplots()
ax.plot(months, station_a_mm, marker="o")
ax.set_title("Monthly rainfall at Station A, Kaski (practice data)")
ax.set_xlabel("Month")
ax.set_ylabel("Rainfall (mm)")

png_path = OUT / "station_a_monthly_rainfall.png"
fig.savefig(png_path, dpi=300, bbox_inches="tight")
plt.show()

print(png_path.exists())
````

`True` confirms the file is in the course `outputs` folder. Open it from the JupyterLab file browser to see it at full size.

## 8. An honest, readable chart: a checklist

A chart can mislead even when every number in it is correct. Before you share a chart, check it against this list:

| Check | Why |
| --- | --- |
| A title that says what, where and, if needed, "practice data" | The reader knows what they are looking at |
| Axis labels **with units** | 8,849 what? Metres or feet? |
| Bars start at **zero** | A bar's length is its value; a cut axis exaggerates differences |
| Shared axes when panels compare the same quantity | Equal heights mean equal values |
| Fonts large enough to read (`fontsize=12` or more) | Charts are often shown small, on slides or phones |
| One colour, unless colour means something | Colour should carry information, not decoration |
| **No rainbow colours** | Rainbows suggest order and steps that are not in the data, and many readers cannot tell red from green |
| A legend whenever there is more than one line or colour | Each colour must be explained |

**Predict:** in the left panel below, the axis starts at 8,000 m. How many times longer will Everest's bar look than Annapurna I's?

````python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11, 4))

ax1.barh(sorted_peaks["name"], sorted_peaks["height_m"], color="grey")
ax1.set_xlim(8000, 8900)     # cutting the axis: do not do this with bars
ax1.set_title("Misleading: axis starts at 8,000 m")
ax1.set_xlabel("Height (m)")

ax2.barh(sorted_peaks["name"], sorted_peaks["height_m"], color="tab:blue")
ax2.set_title("Honest: axis starts at 0 m")
ax2.set_xlabel("Height (m)")

fig.tight_layout()
plt.show()
````

In the left panel, Everest's bar looks about nine times longer than Annapurna I's, yet Everest is less than 10% higher. `ax.set_xlim(start, end)` sets the range of the x-axis; with bars, leave the start at 0.

`color="grey"` or `color="tab:blue"` gives all bars one colour. Matplotlib's named colours include `"tab:blue"`, `"tab:orange"`, `"tab:red"` and `"grey"`.

## Recap

- `fig, ax = plt.subplots()` makes a **figure** (the image) and an **axes** (the plotting area); draw with `ax` methods and finish with `plt.show()`.
- Line charts (`ax.plot`) for ordered sequences, bar charts (`ax.barh`, sorted) to compare amounts, scatter plots (`ax.scatter`) for two numbers per record.
- Every shared chart needs a title, axis labels **with units**, and a legend when there is more than one line or colour.
- `plt.subplots(1, 2, sharey=True)` puts panels side by side on a fair, shared scale. `fig.savefig(path, dpi=300, bbox_inches="tight")` saves, before `plt.show()`.
- Be honest: bars start at zero, colour must mean something, no rainbows, and say when data are practice data.

### Words you learned

| Word | Meaning |
| --- | --- |
| Matplotlib | The Python library for drawing charts; its main tools are loaded as `plt` |
| figure | The whole image, which can hold one or more axes |
| axes | One plotting area inside a figure, with an x-axis and a y-axis |
| tick labels | The numbers or names written along an axis |
| unit | What a number measures in, such as m, ft or mm; written in the axis label |
| line chart | Points joined by lines, for values along an ordered sequence such as months |
| marker | A symbol drawn at each data point, such as `marker="o"` |
| legend | The key that explains which colour or line is which |
| bar chart | Bars whose lengths compare amounts across named things |
| scatter plot | One dot per record, placed by two numbers |
| subplots | Several axes arranged in rows and columns in one figure |
| `sharey` | Makes side-by-side panels use the same y-axis |
| dpi | Dots per inch: the resolution of a saved image; 300 is print quality |
| `bbox_inches="tight"` | Trims empty margins when saving, so labels are not cut off |
| `set_xlim` | Sets the start and end of the x-axis |
| first dimension | The number of values along the first direction of an array or list |
