## 1. From a list of records to a table

In Lesson A4 you stored peaks as a **list of dictionaries**: one dictionary per record. That is a table, but Python cannot show it as one or calculate down its columns.

**pandas** is a library for tables. By custom it is loaded as `import pandas as pd`.

A pandas table is called a **DataFrame**: rows of records and named **columns**, like a spreadsheet. `pd.DataFrame(records)` turns a list of dictionaries into a DataFrame; each key becomes a column.

`pd.read_csv(path)` reads a whole CSV file into a DataFrame. It turns numbers into numbers for you, so no `int()` is needed.

The numbers down the left side (0, 1, 2, …) are the **index**: a label for each row.

````python
import pandas as pd
from pathlib import Path

records = [
    {"name": "Everest", "height_m": 8849, "province": "Koshi"},
    {"name": "Kanchenjunga", "height_m": 8586, "province": "Koshi"},
    {"name": "Dhaulagiri I", "height_m": 8167, "province": "Gandaki"},
]
small = pd.DataFrame(records)
small
````

````python
DATA = Path("..") / "data"
peaks = pd.read_csv(DATA / "nepal_peaks.csv")
peaks
````

## 2. A first look at a table

Before you analyse a table, look at it. These tools answer "what is in here?":

| Code | Shows |
| --- | --- |
| `df.head()` / `df.tail()` | the first / last 5 rows (`head(3)` shows 3) |
| `df.shape` | (rows, columns), as in NumPy |
| `df.columns` | the column names |
| `df.dtypes` | the type of each column |
| `df.info()` | columns, types and how many values are filled in |
| `df.describe()` | count, mean, min, max and more for each number column |

Text columns have the dtype `str`. Older versions of pandas show `object` instead; it means the same here.

## 3. Columns: one column is a Series

`df["height_m"]` picks one column. A single column is a **Series**: a 1-D array of values with the row labels attached.

A Series behaves like a NumPy array. It has `.min()`, `.max()`, `.mean()` and `.sum()`, and arithmetic works on every value at once. `.round(1)` rounds every value.

To pick several columns, put a **list** of names inside the brackets: `df[["name", "height_m"]]`. The outer brackets mean "select", the inner ones make the list. The result is a smaller DataFrame.

**Predict:** what will `peaks["height_m"].mean()` print? You saw it in `describe()` above.

````python
heights = peaks["height_m"]
print(heights)
print(heights.max(), heights.mean())
````

````python
print((peaks["height_m"] * 3.28084).round())
````

````python
peaks[["name", "height_m"]]
````

The Series prints its index on the left and its values on the right, followed by its name and dtype. The mean is 8,380.625 m.

Multiplying the Series by 3.28084 converts all eight heights to feet in one line, as with NumPy.

## 4. Rows: `.iloc` by position, `.loc` by label

There are two ways to pick rows:

- `.iloc[position]` counts positions from 0, like a list or a NumPy array. `peaks.iloc[0]` is the first row and `peaks.iloc[-1]` the last.
- `.loc[label]` uses the row **labels** in the index.

Both take a row and a column: `peaks.iloc[0, 1]` or `by_name.loc["Lhotse", "height_m"]`.

With the default index (0, 1, 2, …) the labels look like positions. `df.set_index("name")` makes a new table with the peak names as row labels, so `.loc` can find a row by name.

A slice of labels, like `.loc["Station A", "jun_mm":"sep_mm"]`, **includes** the end label. That is different from positions, where the end is left out.

````python
print(peaks.iloc[0])      # the first row, as a Series
peaks.iloc[0:3]           # rows at positions 0, 1, 2
````

````python
by_name = peaks.set_index("name")
print(by_name.loc["Lhotse", "height_m"])
by_name.loc["Makalu"]
````

````python
stations = rain.set_index("station")
stations.loc["Station A", "jan_mm":"dec_mm"]
````

A single row comes out as a Series, with the column names as its labels. `by_name.loc["Lhotse", "height_m"]` gives 8,516.

The last cell shows the twelve months of Station A, from 35 mm in January to 1,045 mm in July.

## 5. Filtering rows with conditions

A comparison on a column gives a Series of `True` and `False`: a boolean mask, as in Lesson A8.

Put the mask inside square brackets to keep only the rows where it is `True`: `peaks[peaks["height_m"] > 8500]`.

Combine conditions with `&` (and) and `|` (or), with **brackets around each condition**. Use `==` to compare text: `slides["trigger"] == "earthquake"`.

`.loc[mask, "name"]` filters the rows and picks a column in one step.

**Predict:** which peaks are higher than 8,500 m?

````python
peaks[peaks["height_m"] > 8500]
````

````python
high_gandaki = (peaks["province"] == "Gandaki") & (peaks["height_m"] > 8150)
peaks[high_gandaki]
````

````python
early = peaks["first_ascent"] < 1955
print(peaks.loc[early, "name"])
````

Three peaks are above 8,500 m: Everest, Kanchenjunga and Lhotse. The rows keep their original index labels (0, 1, 2).

In Gandaki province, two peaks are above 8,150 m: Dhaulagiri I and Manaslu. Three peaks were first climbed before 1955: Everest, Cho Oyu and Annapurna I.

## 6. Sorting and adding new columns

`df.sort_values("height_m")` sorts the rows from smallest to largest. Add `ascending=False` for largest first. Each row keeps its index label, so you can see where it came from.

Sorting gives back a **new** table and leaves the original as it was. Store the result in a variable if you want to keep it.

To add a column, assign to a new name: `peaks["height_ft"] = peaks["height_m"] * 3.28084`. The calculation runs for every row.

To add up several columns in each row, select them and use `.sum(axis=1)`, which works across each row as in A8. Here that turns twelve months into one **annual** total per station.

````python
peaks["height_ft"] = (peaks["height_m"] * 3.28084).round()
peaks[["name", "height_m", "height_ft"]]
````

````python
rain["annual_mm"] = rain.loc[:, "jan_mm":"dec_mm"].sum(axis=1)
wettest_first = rain.sort_values("annual_mm", ascending=False)
wettest_first[["station", "district", "elevation_m", "annual_mm"]]
````

`rain.loc[:, "jan_mm":"dec_mm"]` means "all rows, columns from January to December".

Station A (Kaski) is the wettest with 3,936 mm a year, and Station C (Mustang) the driest with 273 mm. Notice the index labels: Station A keeps label 0 and Station C keeps label 2.

## 7. Counting and grouping

`df["province"].value_counts()` counts how often each value appears, most common first.

`df.groupby("province")` splits the table into groups that share a value. You then pick a column and a summary:

`peaks.groupby("province")["height_m"].mean()`

Read it as: *split the peaks by province, take the heights, and average each group.* `.sum()`, `.count()`, `.min()` and `.max()` work the same way. This "split, then summarise" pattern answers a very large share of everyday data questions.

**Predict:** which province has the higher mean peak height, Koshi or Gandaki?

````python
print(peaks["province"].value_counts())
````

````python
print(peaks.groupby("province")["height_m"].mean())
````

````python
print(slides.groupby("trigger")["area_m2"].count())
print(slides.groupby("trigger")["area_m2"].mean().round())
````

Koshi has 5 of the eight peaks and Gandaki 3. Koshi's peaks average 8,524.8 m and Gandaki's 8,140.3 m.

In the landslide records, 29 are triggered by rainfall, 8 by road cuts and 3 by earthquakes. Road-cut landslides are the smallest on average, about 1,888 m². The groups come out in alphabetical order.

## 8. Saving a table to a file

`df.to_csv(path, index=False)` writes a DataFrame to a CSV file. `index=False` leaves out the row numbers, which are not real data.

Save your results in the course `outputs` folder, as in Lesson A7. `mkdir(exist_ok=True)` creates it if it is missing, and does nothing if it is already there.

Reading the file back is a good habit: it proves the file holds what you think it holds. Spreadsheet programs can open the CSV files you write.

````python
OUT = Path("..") / "outputs"
OUT.mkdir(exist_ok=True)

summary = rain[["station", "district", "elevation_m", "annual_mm"]]
summary.to_csv(OUT / "rain_annual.csv", index=False)

print((OUT / "rain_annual.csv").exists())
print((OUT / "rain_annual.csv").read_text())
````

The file has a header line and one line per station, with no index column.

## Recap

- `import pandas as pd`; `pd.read_csv(path)` loads a CSV file into a **DataFrame**, a table with named columns and an index.
- Look before you analyse: `head()`, `shape`, `columns`, `dtypes`, `info()` and `describe()`.
- `df["col"]` gives a **Series**; `df[["a", "b"]]` gives several columns. `.iloc` picks rows by position and `.loc` by label.
- Filter with masks (`&` for and, `|` for or, brackets around each condition), sort with `sort_values`, and add columns with `df["new"] = ...`.
- `value_counts()` counts, `groupby("col")["other"].mean()` summarises each group, and `to_csv(path, index=False)` saves the result.
