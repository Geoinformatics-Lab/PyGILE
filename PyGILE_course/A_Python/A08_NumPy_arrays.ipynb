## 1. Why arrays: one calculation for many numbers

In Lesson A3 you stored the heights of the eight peaks in a list. To convert them all to feet, you needed a loop.

**NumPy** is a **library**: a package of ready-made tools that you load with `import`. By custom it is loaded as `import numpy as np`, so you can write the short name `np`.

`np.array(a_list)` turns a list into an **array**. An array looks like a list, but every value has the same type. Arithmetic on an array is applied to **every value at once**, with no loop. This is called **vectorised** arithmetic.

**Predict:** in the next cell, what will `heights_list * 2` print, and what will `heights_m * 2` print? Write down your guess first.

## 2. A mountain as a grid: 2-D arrays

A map of heights can be stored as a grid. Each **cell** of the grid holds the elevation of one square of ground. A grid of elevations is called a **DEM** (digital elevation model). You will work with real DEMs later in the course.

You build a grid from a **list of lists**: each inner list is one **row**. An array with rows and columns is a **2-D array** (two-dimensional).

The grid below is a small made-up mountain: 5 rows by 6 columns, each cell 500 m × 500 m (made-up values for practice). Row 0 is the northern edge, column 0 the western edge.

Every array carries some facts about itself, called **attributes**. You read them with a dot and **no brackets**:

| Attribute | Meaning |
| --- | --- |
| `.shape` | the size in each direction: (rows, columns) |
| `.ndim` | the number of dimensions: 1 for a row of values, 2 for a grid |
| `.size` | the total number of values |
| `.dtype` | the type of the values: `int64` for whole numbers, `float64` for decimals |

**Predict:** what will `elev_m.shape` be?

````python
elev_m = np.array([
    [4210, 4350, 4480, 4520, 4400, 4290],   # row 0 (north edge)
    [4380, 4620, 4890, 4950, 4700, 4460],   # row 1
    [4510, 4870, 5230, 5410, 5020, 4630],   # row 2
    [4470, 4790, 5150, 5290, 4980, 4590],   # row 3
    [4300, 4520, 4760, 4810, 4650, 4410],   # row 4 (south edge)
])
cell_size_m = 500   # each cell covers 500 m x 500 m of ground

print(elev_m)
````

````python
print(elev_m.shape)
print(elev_m.ndim)
print(elev_m.size)
print(elev_m.dtype)
````

The shape `(5, 6)` means 5 rows and 6 columns. Round brackets with values inside, like `(5, 6)`, are a **tuple**: a list that cannot be changed. You read a tuple with an index, like a list: `elev_m.shape[0]` is 5.

`int64` means whole numbers stored in 64 bits of memory. The `heights_km` array from step 1 holds decimals, so its dtype is `float64`.

## 3. Picking one cell: `[row, column]`

To read one cell, write the array name, then square brackets with the **row** first and the **column** second: `elev_m[2, 3]`.

Both counts start at 0, as with lists. Rows count down from the top (north); columns count across from the left (west). Negative numbers count from the end: `elev_m[-1, -1]` is the bottom-right corner.

Picking values by position is called **indexing**, and the numbers in the brackets are **indices**.

**Predict:** look at the printed grid above. What is `elev_m[1, 2]`?

````python
print(elev_m[0, 0])    # top-left corner
print(elev_m[1, 2])    # row 1, column 2
print(elev_m[2, 3])
print(elev_m[-1, -1])  # bottom-right corner
````

`elev_m[1, 2]` is 4,890 m: go down to row 1, then across to column 2. The cell `[2, 3]`, at 5,410 m, is the summit of our made-up mountain.

## 4. Slicing: rows, columns and blocks

You can cut out many cells at once with a **slice**, `start:stop`, as with lists in A3. The `stop` position is **not** included.

A colon on its own, `:`, means "all of them". So:

- `elev_m[2, :]` is all of row 2,
- `elev_m[:, 3]` is all of column 3,
- `elev_m[1:4, 2:5]` is rows 1 to 3 and columns 2 to 4: a 3 × 3 **block**.

A slice is itself an array, so it has a shape, and you can calculate with it.

````python
print(elev_m[2, :])        # row 2
print(elev_m[:, 3])        # column 3
block = elev_m[1:4, 2:5]   # rows 1-3, columns 2-4
print(block)
print(block.shape)
````

A whole row or column comes out as a 1-D array. The block keeps two dimensions, with shape `(3, 3)`, and its centre is the summit.

## 5. Arithmetic on a whole grid

Arithmetic on a 2-D array works on every cell, as it did on the 1-D array in step 1. `elev_m * 3.28084` converts the whole grid to feet in one line.

When you combine a grid with a single number, NumPy stretches that number across every cell. NumPy calls this **broadcasting**.

Two arrays of the **same shape** can be added or subtracted cell by cell. This is how you compare two surveys of the same area.

`np.round(array, 1)` rounds every value to one decimal place. With no second number it rounds to whole numbers but keeps the `float` type, so `13812.0` is printed as `13812.`.

````python
elev_ft = elev_m * 3.28084
print(np.round(elev_ft, 1))
````

````python
# two small surveys of the same 2 x 3 area (made-up values for practice)
survey_2020_m = np.array([[1510, 1522, 1530],
                          [1498, 1505, 1519]])
survey_2024_m = np.array([[1510, 1517, 1530],
                          [1490, 1505, 1519]])
change_m = survey_2024_m - survey_2020_m
print(change_m)
````

The first cell shows the whole grid in feet; the summit is about 17,749.3 ft.

In the second cell, most cells did not change (0). Two cells lost height: −5 m and −8 m. On a real slope that pattern could mean material moved downhill. You will meet real change maps much later in the course.

## 6. Summaries: mean, min, max, and where the top is

Arrays have **methods**, actions you call with a dot and brackets: `.min()`, `.max()`, `.mean()` and `.sum()`. Each one summarises the whole array into one number.

Add `axis=1` to summarise **across each row** (one answer per row), or `axis=0` to summarise **down each column** (one answer per column).

`.argmax()` tells you **where** the maximum is, and `.argmin()` where the minimum is. On a 2-D array the answer is a single position, counted cell by cell along row 0, then row 1, and so on. `np.unravel_index(position, elev_m.shape)` turns that number back into a row and a column.

````python
print(elev_m.min(), elev_m.max())
print(round(elev_m.mean(), 1))
print(np.round(elev_m.mean(axis=1), 1))   # one mean per row
````

````python
position = elev_m.argmax()
print(position)
row, col = np.unravel_index(position, elev_m.shape)
print(row, col)
print(elev_m[row, col])
````

The grid runs from 4,210 m to 5,410 m, with a mean of 4,688 m. Row 2 has the highest mean, 4,945 m: it crosses the summit.

`argmax()` gave 15: counting cells row by row, row 0 holds positions 0 to 5, row 1 holds 6 to 11, and row 2 starts at 12. Position 15 is therefore row 2, column 3.

`np.unravel_index` gives back two numbers. Writing `row, col = ...` stores them in two names at once, as `for name, height in ...` did in Lesson A4.

## 7. Masks: asking a question of every cell

A comparison on an array answers the question for every cell, giving an array of `True` and `False` with the same shape. This is called a **boolean mask**.

`mask.sum()` counts the `True` cells, because `True` counts as 1 and `False` as 0. `elev_m[mask]` keeps only the values where the mask is `True`, as a 1-D array.

To combine two conditions, use `&` for "and" and `|` for "or", with **brackets around each condition**: `(elev_m > 4500) & (elev_m < 5000)`. The words `and` and `or` from A4 do not work on arrays.

`np.where(condition, value_if_true, value_if_false)` builds a new grid, choosing a value for each cell.

**Predict:** how many cells of `elev_m` are above 5,000 m? Count them in the printed grid before you run the cell.

````python
high = elev_m > 5000
print(high)
print(high.sum())
print(elev_m[high])
````

````python
# label each cell, using 5,000 m as a made-up snow line for practice
cover = np.where(elev_m > 5000, "snow", "rock")
print(cover)
````

## 8. Differences between neighbours: a first look at slope

Slope is how much the height changes over a horizontal distance. On a grid, you compare each cell with its neighbour.

`np.diff(values)` subtracts each value from the next one. Six values give five differences. A positive difference means uphill, a negative one downhill.

On a 2-D grid, `np.diff(elev_m, axis=1)` works along each row (west to east), and `axis=0` down each column (north to south).

Dividing the height difference by the cell size gives rise ÷ run, as in Lesson A1. Multiply by 100 for a gradient in percent. NumPy has array versions of the `math` functions from A5, such as `np.arctan` and `np.degrees`, plus `np.abs` to drop the minus sign.

````python
row_2_m = elev_m[2]
steps_m = np.diff(row_2_m)
print(row_2_m)
print(steps_m)
print(steps_m / cell_size_m * 100)   # gradient in percent
````

````python
east_west_steps_m = np.diff(elev_m, axis=1)
print(east_west_steps_m.shape)
````

Along row 2 the ground rises 360, 360 and 180 m from cell to cell, then drops 390 m twice. Over 500 m cells, those are gradients of 72%, 72%, 36%, −78% and −78%.

The differences along the rows have shape `(5, 5)`: each row of 6 cells gives 5 differences.

## 10. A quiet mistake: numbers that wrap around

Some errors give no message at all. Each dtype has a limit. The type `uint8` stores whole numbers from 0 to 255 only, in one byte of memory. Satellite images often use it to save space.

If a calculation goes past 255, the value **wraps around** to 0 and keeps counting, with no warning. **Predict:** what is 200 + 100 in `uint8`?

````python
brightness = np.array([200, 100, 250], dtype=np.uint8)
print(brightness + 100)
````

200 + 100 = 300, which wrapped round to 44 (300 − 256). 250 + 100 became 94. Only 100 + 100 = 200 is correct, because it stays under 255.

The fix is to change the dtype before you calculate. `.astype(float)` makes a copy of the array with decimal numbers, which have plenty of room:

````python
brightness_f = brightness.astype(float)
print(brightness_f + 100)
````

Now the results are 300, 200 and 350. Whenever you work with image values, check `.dtype` first and convert to `float` before arithmetic.

## Recap

- `import numpy as np`, then `np.array(a_list)` makes an **array**. Arithmetic on an array works on every value at once.
- A 2-D array is a grid: `.shape` gives (rows, columns). Pick cells with `grid[row, col]` and cut rows, columns or blocks with slices like `grid[1:4, 2:5]`.
- `.min()`, `.max()`, `.mean()` and `.sum()` summarise a grid; `axis=0` works down columns and `axis=1` across rows. `argmax()` with `np.unravel_index` finds where the top is.
- A comparison gives a **boolean mask**. Use it to count (`mask.sum()`), to select (`grid[mask]`) and to build new grids with `np.where`. Combine masks with `&` and `|`.
- Check `.shape` when you see `could not be broadcast`, and check `.dtype` before arithmetic: `uint8` wraps around past 255, so convert with `.astype(float)`.
